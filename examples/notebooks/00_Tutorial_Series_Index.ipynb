# EPyR Tools Tutorial Series

**EPyR Tools** is a Python package for analyzing Electron Paramagnetic Resonance
(EPR) spectroscopy data from Bruker spectrometers. It loads BES3T (`.DTA`/`.DSC`)
and ESP/WinEPR (`.spc`/`.par`) files and provides baseline correction, lineshape
and relaxation fitting, FFT-based signal processing, FAIR-compliant export, and
EPR-specific physical constants.

This series is a hands-on tour of the public API. Each notebook is standalone,
uses real experimental data from `../data/`, and runs top to bottom.

## Notebooks

| # | Notebook | Covers |
|---|----------|--------|
| 01 | `01_Loading_and_Plotting` | `eprload`, parameter inspection, 1D/2D plotting |
| 02 | `02_Baseline_Correction` | Polynomial, automatic, and exponential baselines |
| 03 | `03_Lineshape_Analysis_and_Fitting` | Gaussian/Lorentzian/Voigt, derivatives, curve fitting |
| 04 | `04_Relaxation_Fitting` | T1/T2 decay/recovery models (new in v0.4.0) |
| 05 | `05_Signal_Processing_and_FFT` | Frequency analysis of Rabi oscillations, apodization |
| 06 | `06_FAIR_Conversion_and_Export` | CSV/JSON/HDF5 export and validation |
| 07 | `07_Physics_Units_and_Constants` | CODATA constants, field/frequency conversions |
