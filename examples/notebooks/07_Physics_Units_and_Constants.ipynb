# 07 - Physics, Units, and Constants

`epyr.physics` provides CODATA 2022 constants and EPR-specific unit conversions,
all available at the top level (`epyr.GFREE`, `epyr.unitconvert`, ...).

```python
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")  # keep tutorial output readable

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

import epyr

DATA = Path("..") / "data"   # example datasets, relative to this notebook
print("EPyR Tools version:", epyr.__version__)
```

## Physical constants (SI)

```python
print("free-electron g-factor :", epyr.GFREE)
print("Bohr magneton  (J/T)   :", epyr.BMAGN)
print("Planck constant (J s)  :", epyr.PLANCK)
print("Boltzmann const (J/K)  :", epyr.BOLTZM)
print("nuclear magneton (J/T) :", epyr.NMAGN)
```

## Unit conversions

`unitconvert(value, "from->to")` handles the units common in EPR: `mT`, `MHz`,
`cm^-1`, `eV`, `K`. Dedicated helpers exist too (`mt_to_mhz`, `mhz_to_mt`, ...).

```python
from epyr.physics import mt_to_mhz, mhz_to_mt, cm_inv_to_mhz

print("9500 MHz ->", epyr.unitconvert(9500.0, "MHz->mT"), "mT (at g_free)")
print("340  mT  ->", mt_to_mhz(340.0), "MHz")
print("1.0  cm^-1 ->", cm_inv_to_mhz(1.0), "MHz")
```

## Field, frequency, and the g-factor

The resonance condition is h*nu = g * muB * B. Below, a known field and frequency
recover g near the free-electron value.

```python
from epyr.physics import magnetic_field_to_frequency, frequency_to_magnetic_field

nu = 9.5e9          # Hz (X-band)
B_tesla = 0.339     # T, near g = 2

g = epyr.PLANCK * nu / (epyr.BMAGN * B_tesla)
print(f"g-factor from nu={nu/1e9:.3f} GHz, B={B_tesla*1e3:.1f} mT  ->  g = {g:.4f}")

print("field for g_free at 9.5 GHz:",
      frequency_to_magnetic_field(nu) * 1e3, "mT")
```
